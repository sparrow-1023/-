```python
import random

# 1. 冒險者登入
name = input("請輸入你的冒險者名字：")

# 裝備選擇選單
print("\n請選擇你的使用裝備：")
print("1. 阿姆斯特朗旋風噴射阿姆斯特朗砲 (25% 機率炸掉對手贏的回合)")
print("2. 壽來 (每隔一回合使電腦出布的機率改為 50%)")
print("3. 我要採一朵小花送給媽媽 (當前一回合輸了，能看到電腦這回合要出甚麼)")

while True:
    eq_choice = input("請輸入裝備編號 (1 / 2 / 3)：").strip()
    if eq_choice == "1":
        equipment = "阿姆斯特朗旋風噴射阿姆斯特朗砲"
        break
    elif eq_choice == "2":
        equipment = "壽來"
        break
    elif eq_choice == "3":
        equipment = "我要採一朵小花送給媽媽"
        break
    else:
        print("⚠️ 指令錯誤，請輸入 1、2 或 3！")

# 印出前置劇情
print("\n=== 系統廣播 ===")
print(f"歡迎勇敢的冒險者【{name}】！")
print(f"你已經裝備了【{equipment}】，準備好迎接第一關的 【猜拳】 挑戰！")


print("\n\n=== ✊✋✌️ 剪刀石頭布 終極大對決 ===")
print("規則：先獲得 3 勝的一方就是最終贏家！\n")

player_wins = 0
computer_wins = 0
round_num = 1
last_round_lost = False  # 記錄上一回合是否輸掉（給裝備 3 使用）

options = ["剪刀", "石頭", "布"]

# 2. 遊戲主迴圈
while player_wins < 3 and computer_wins < 3:
    print(f"--- 第 {round_num} 回合 ---")
    print(f"🏆 目前戰況 - 你：{player_wins} 勝 | 電腦：{computer_wins} 勝")

    # --- 電腦出拳邏輯 ---
    # 裝備 2 效果：每隔一回合（如第 1, 3, 5 回合），電腦出「布」的機率變為 50%
    if eq_choice == "2" and round_num % 2 == 1:
        computer_choice = random.choices(
            ["剪刀", "石頭", "布"], weights=[25, 25, 50], k=1
        )[0]
    else:
        computer_choice = random.choice(options)

    # --- 裝備 3 效果：若前一回合輸了，預先顯示電腦本回合出的拳 ---
    if eq_choice == "3" and last_round_lost:
        print(
            f"🌸【我要採一朵小花送給媽媽】技能發動！你預知到了電腦這回合要出：【{computer_choice}】"
        )

    # 3. 玩家出拳
    player_choice = input("請出拳 (輸入 剪刀、石頭 或 布)：").strip()

    if (
        player_choice != "剪刀"
        and player_choice != "石頭"
        and player_choice != "布"
    ):
        print("⚠️ 哎呀！指令錯誤，請輸入 剪刀、石頭 或 布。\n")
        continue

    print(f"🤖 電腦出了：{computer_choice}")

    # 4. 勝負判斷
    if player_choice == computer_choice:
        print("平手！沒有人得分。\n")
        last_round_lost = False
    elif (
        (player_choice == "剪刀" and computer_choice == "布")
        or (player_choice == "石頭" and computer_choice == "剪刀")
        or (player_choice == "布" and computer_choice == "石頭")
    ):
        print("🎉 這回合你贏了！\n")
        player_wins += 1
        last_round_lost = False
    else:
        # 電腦原本贏得這回合，判定裝備 1 觸發
        if eq_choice == "1" and random.random() < 0.25:
            print(
                "🚀【阿姆斯特朗旋風噴射阿姆斯特朗砲】發動！轟飛了電腦這回合的勝利！電腦無法得分！\n"
            )
            last_round_lost = False
        else:
            print("💥 這回合電腦贏了！\n")
            computer_wins += 1
            last_round_lost = True

    round_num += 1


# 5. 最終結算
print("=== 🌟 最終結果 🌟 ===")

if player_wins == 3:
    print(
        f"🏆 恭喜你以 {player_wins} : {computer_wins}"
        " 擊敗了電腦，成為剪刀石頭布大師！"
    )
else:
    print(
        f"💻 很可惜，電腦以 {computer_wins} : {player_wins}"
        " 獲得了最終勝利，下次再來挑戰吧！"
    )
```
